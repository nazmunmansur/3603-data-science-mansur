## 🎛️ Default Parameter Values

**Problem.** Here's `count_above` from Module 03b. In this class, the cutoff is almost always **70** (passing):

````python
def count_above(numbers, threshold):
    count = 0
    for x in numbers:
        if x > threshold:
            count += 1
    return count

scores = [87, 91, 64, 88, 55]
print(count_above(scores, 70))
print(count_above(scores, 70))
print(count_above(scores, 70))
print(count_above(scores, 90))   # only occasionally something else
````

We keep typing `70` over and over. A **default value** means: *use this unless the caller says otherwise.*

````python
def count_above(numbers, threshold=70):
    count = 0
    for x in numbers:
        if x > threshold:
            count += 1
    return count
````

````python
print(count_above(scores)) # 3 as 70 is threshold by default
print(count_above(scores, 90)) # 1
print(count_above(scores, 50)) # 5
````

You've seen this idea before, with `dict.get(key, default)`: *use this value unless something more specific is available.*

````python
student = {"name": "Ada"}
print(student.get("major", "Undeclared"))    # dictionary default

def describe(name, major="Undeclared"):      # function default
    return f"{name} — {major}"

print(describe("Ada"))
print(describe("Ada", "Data Science"))
````

**Rule:** parameters with defaults go **after** parameters without defaults. `def f(a=1, b):` is a `SyntaxError`, because Python couldn't tell which argument goes where.

## 🏷️ Keyword Arguments

**Problem.** What do the `True` and `False` mean in this call?

```python
report(students, 90, True, False)
```

You can't tell without opening up the function. **Keyword arguments** put the parameter's name in the call:

```python
report(students, cutoff=90, show_names=True, sort=False)
```

The code does the same thing, but now the call explains itself. Arguments passed by name can also be given **in any order**.

````python
def describe(name, major="Undeclared"):
    return f"{name} — {major}"

print(describe("Ada", "Math"))               # positional: matched by order
print(describe(name="Ada", major="Math"))    # keyword: matched by name
print(describe(major="Math", name="Ada"))    # keyword: order doesn't matter
````

## 📦 Returning More Than One Value

**Problem.** You want both the lowest and the highest score. Two functions that each loop over the list is wasteful, and a function can only `return` **once**.

The fix is to return **one tuple** that has two parts. (Remember: a tuple is *one thing with several fixed parts*.)

````python
def low_high(numbers):
    lo = numbers[0]
    hi = numbers[0]
    for n in numbers:
        if n < lo:
            lo = n
        if n > hi:
            hi = n
    return lo, hi          # the comma makes a tuple: (lo, hi)

result = low_high([87, 91, 64, 88, 55])
print(result)
print(type(result))

lo, hi = low_high([87, 91, 64, 88, 55])   # unpack right away
print("lowest:", lo, "highest:", hi)
````

```text
                  ┌──────────┐
 [87, 91, …] ───► │ low_high │ ───► (55, 91) ───► lo = 55
                  └──────────┘                    hi = 91
```

## 🏠 Scope — What Happens Inside Stays Inside

**Problem.** Predict what this prints:

````python
x = 10

def change_x():
    x = 99
    print("inside: ", x)

change_x()
print("outside:", x)
````

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_
* inside: 99
* outside: 10

The `x = 99` inside the function creates a **new, local** `x` that only exists while the function runs. The outer `x` (the **global** one) is never touched.

It works the other way too: a variable created inside a function **does not exist** outside it.

````python
# ⚠️ This cell raises an error on purpose. Read the error message!
def make_total(numbers):
    total = 0
    for n in numbers:
        total += n
    return total

make_total([1, 2, 3])
print(total)
````

### 🧠 Mental Model: A Room With Two Doors

```text
               ┌─────────────────────────┐
 parameters ──►│     function's room     │──► return
 (the way in)  │  local variables live   │   (the way out)
               │  here and disappear     │
               │  when the call ends     │
               └─────────────────────────┘
```

A well-written function gets data **only** through its parameters and gives data back **only** through `return`.

Remember **Patient #4** from the Broken Function Clinic? Its `average(numbers)` ignored its parameter and read the global `scores` instead. Python allowed it, but it's exactly the bug this rule prevents. If a function needs a value, **make it a parameter**.

### Practice 3 — Return a Tuple (harder)
Write `summarize(numbers)` that returns `(total, average, highest)`. Unpack the result into three variables and print each one.

For `[10, 20, 30, 40]`, the result should be `(100, 25.0, 40)`.

````python
def summarize(numbers):
    # Your code here
    total = sum(numbers)
    average = total / len(numbers)
    highest = max(numbers)
    return total, average, highest
    
values = [10, 20, 30, 40]
    
 # Call the function
result = summarize(values)
print(result)
    
# Unpack the tuple
total, average, highest = summarize(values)
    
print(total)
print(average)
print(highest)
````

## 🔥 Challenge (Optional) — The Mutable Default Gotcha

A default value is created **once**, when the function is *defined*, not each time it's called. With a list as the default, that causes a well-known bug:

```python
def add_item(item, cart=[]):
    cart.append(item)
    return cart

print(add_item("apple"))
print(add_item("banana"))   # you'd expect ['banana'], but...
```

Run it below, then fix it: make the default `cart=None`, and inside the function, create a new empty list `if cart is None`.

*Seen in the wild:* in documentation you'll sometimes see `def f(*args, **kwargs)`. That means "accepts any number of positional / keyword arguments." You only need to be able to **recognize** it for now.

````python
def add_item(item, cart=None):

    if cart is None:
        cart = []

    cart.append(item)
    return cart

print(add_item("apple"))
print(add_item("banana"))

# Your fixed version here
````
