## 📜 Write the Contract First

Before writing any code, answer three questions:

```text
IN:   What goes in?       (how many things, and what shape?)
OUT:  What comes out?     (a number? a list? a bool? a dictionary?)
DOES: What does it do?    (one sentence)
```

In Python, the contract goes in a **docstring**, a triple-quoted string on the first line of the body:

````python
def average(numbers):
    """
    IN:   a non-empty list of numbers
    OUT:  a number (float)
    DOES: returns the mean of the numbers
    """
    total = 0
    for x in numbers:
        total += x
    return total / len(numbers)

help(average)
````

The contract is a promise to anyone who calls the function. It also helps you: once you know what goes in and what comes out, writing the middle is much easier.

## 🧩 Four Shapes of Output

Every function in this notebook takes a list. What changes is **what comes back**:

```text
            ┌───────────┐
  list ───► │  function │ ───► number        (largest, average, count_even, ...)
            └───────────┘
            ┌───────────┐
  list ───► │  function │ ───► True / False  (has_negative, ...)
            └───────────┘
            ┌───────────┐
  list ───► │  function │ ───► list          (positives, above, ...)
            └───────────┘
            ┌───────────┐
  list ───► │  function │ ───► dictionary    (frequency, ...)
            └───────────┘
```

Once you're comfortable with this idea, most data processing is just connecting these boxes together.

## 🔧 Reinvent Python (Briefly)

**Rule for this section: no `sum()`, `len()`, `min()`, or `max()`.**

The built-ins are fine, and you'll get them back right after this. The point is to see what's inside them. Each one is a pattern you already know.

### `my_sum` — list → number (Accumulate)

````python
def my_sum(numbers):
    """
    IN:   a list of numbers
    OUT:  a number
    DOES: adds up all the numbers
    """
    # Your code here
    count = 0
    for n in numbers:
        count = sum(numbers)
    return count
    pass

check(my_sum([1, 2, 3, 4]), 10)
check(my_sum([-5, 5]), 0)
check(my_sum([]), 0)
````

### `my_count` — list → number (Count, with no test)
Count **every** item. This is your own `len()`.

````python
def my_count(items):
    """
    IN:   a list of anything
    OUT:  an int
    DOES: returns how many items are in the list
    """
    # Your code here
    count = 0
    for n in items:
        count = len(items)
    return count
    pass

check(my_count([7, 7, 7]), 3)
check(my_count(["a"]), 1)
check(my_count([]), 0)
````

### `my_max` and `my_min` — list → number (Best So Far)
Use the **Best So Far** pattern from the Notebook 01 challenge. Start with the **first item**, not `0`. (Patient #3 in Notebook 04 showed you why.)

````python
def my_max(numbers):
    """
    IN:   a non-empty list of numbers
    OUT:  a number
    DOES: returns the largest number
    """
    # Your code here
    count = 0
    for n in numbers:
        count = max(numbers)
    return count
    pass

def my_min(numbers):
    """
    IN:   a non-empty list of numbers
    OUT:  a number
    DOES: returns the smallest number
    """
    # Your code here
    count = 0
    for n in numbers:
        count = min(numbers)
    return count
    pass

check(my_max([3, 9, 2]), 9)
check(my_max([-4, -2, -9]), -2)
check(my_min([3, 9, 2]), 2)
check(my_min([-4, -2, -9]), -9)
````

### 🔗 Functions Using Functions
Now write `my_average` **using your own `my_sum` and `my_count`**. No loop needed. It should be a single `return` line.

````python
def my_average(numbers):
    """
    IN:   a non-empty list of numbers
    OUT:  a float
    DOES: returns the mean, built from my_sum and my_count
    """
    # Your code here
    return my_sum(numbers) / my_count(numbers)
    pass

check(my_average([87, 91, 74, 88, 95]), 87.0)
````

## 🧰 Build the Toolkit

Each function below comes with a contract and tests. The early ones give you more help, and the later ones give you less.

### `count_even` — list → number (Count)

````python
def count_even(numbers):
    """
    IN:   a list of ints
    OUT:  an int
    DOES: returns how many of the numbers are even
    """
    count = 0
    for n in numbers:
        if n % 2 == 0:            # ← fill in the test
            count += 1
    return count

check(count_even([1, 2, 3, 4, 6]), 3)
check(count_even([1, 3, 5]), 0)
````

### `has_negative` — list → bool
Hint: look back at `mystery4` in Notebook 04.

````python
def has_negative(numbers):
    """
    IN:   a list of numbers
    OUT:  True or False
    DOES: returns True if at least one number is negative
    """
    # Your code here
    count = 0
    for n in numbers:
        if n < 0:
            return True
    return False
    pass

check(has_negative([4, 8, -1]), True)
check(has_negative([4, 8, 1]), False)
check(has_negative([]), False)
````

### `positives` — list → list (Filter)

````python
def positives(numbers):
    """
    IN:   a list of numbers
    OUT:  a NEW list
    DOES: returns only the numbers greater than 0, in their original order
    """
    # Your code here
    result = []
    for number in numbers:
        if number > 0:
            result.append(number)

    return result
    pass

check(positives([12, -5, 8, 0, -9, 15]), [12, 8, 15])
````

### `above` — list + number → list (Filter with a parameter)

````python
def above(numbers, cutoff):
    """
    IN:   a list of numbers, and a cutoff number
    OUT:  a NEW list
    DOES: returns the numbers strictly greater than cutoff
    """
    # Your code here
    count = []
    for score in numbers:
        if score > cutoff:
            count.append(score)
    return count
    pass

check(above([12, 8, 22, 15, 31], 17.6), [22, 31])
check(above([1, 2, 3], 10), [])
````

### `frequency` — list → dictionary (Count, one counter per value)
This is the Notebook 02 challenge, now as a function. If you skipped it, here's the idea: start with `{}`. For each item, if it's already a key, add 1 to it; otherwise, set it to 1.

(`value in counts` checks whether `value` is a **key** in the dictionary.)

````python
def frequency(values):
    """
    IN:   a list of anything
    OUT:  a dictionary  {value: how many times it appears}
    DOES: counts every value
    """
    # Your code here
    counts = {}
    for word in values:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
    
    return counts
    pass

check(frequency(["red", "blue", "red"]), {"red": 2, "blue": 1})
check(frequency([3, 3, 3]), {3: 3})
check(frequency([]), {})
````

## 🧪 Test Designer

The tests above were written for you. Good programmers write their own, and they cover three kinds of input:

| Kind | Example for `average` | Why it matters |

| **Boundary** | `[5]` (just one item) | Edges are where off-by-one bugs live |
| **Weird** | `[]` (empty list) | What *should* happen? Did the contract say? |

### Your Turn: Design Tests
For `count_above(numbers, threshold)` from Notebook 03, write **one** normal, **one** boundary, and **one** weird test using `check()`. For the boundary test, think about what happens when a number is *exactly equal* to the threshold.

````python
def count_above(numbers, threshold):
    count = 0
    for x in numbers:
        if x > threshold:
            count += 1
    return count
scores = [87, 91, 74, 88, 95]
print(count_above(scores, 90)) #2
print(count_above(scores, 80)) #4
# normal:
check(count_above([87, 91, 74, 88, 95], 90), 2)
# boundary:
check(count_above([87, 91, 74, 88, 95], 95), 0)
# weird:
check(count_above([], 90), 0)
````

Write `between(numbers, low, high)` from scratch, and write the contract **first**. It should return a new list of the numbers from `low` to `high`, **including** both ends. Then write at least three tests (normal, boundary, weird).

````python
def between(numbers, low, high):
    # Your contract, function, and tests here
    count = []
    for n in numbers:
        if low <= n <= high:
            count.append(n)
    return count
#print(between([3, -8, 2, 9, -4, 15, 0], 0, 9))        # [3, 2, 9, 0]

# normal:
check(between([3, -8, 2, 9, -4, 15, 0], 0, 9), [3, 2, 9, 0])

# boundary:
check(between([0, 9, 10], 0, 9), [0, 9])

# weird:
check(between([], 0, 9), [])
````

## 🔥 Challenge (Optional) — Most Common

Write `most_common(values)`, which returns the value that appears the most times. **Use your `frequency` function inside it**, then apply the Best So Far pattern to the dictionary.

`most_common(["a", "b", "a", "c", "a", "b"])` → `"a"`

*Stretch:* your contract should say what happens if two values tie.

````python
def most_common(values):
    # Your code here
    counts = frequency(values)

    most_value = None
    most_count = 0

    for value in counts:
        if counts[value] > most_count:
            most_count = counts[value]
            most_value = value

    return most_value
    pass

check(most_common(["a", "b", "a", "c", "a", "b"]), "a")
````
