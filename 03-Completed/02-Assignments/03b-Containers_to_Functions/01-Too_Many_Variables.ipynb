## 🔁 Doing Something With Every Item

Having one name for the whole group only pays off if you can handle **every** item without naming each one. That's what a `for` loop does.

````python
scores = [87, 91, 74, 88, 95]

for score in scores:
    print(score)
````

Read it in plain English: *"For each score in scores, print that score."*

`score` (singular) is a name that points at **one item at a time**. Python gives it the first item, runs the indented body, gives it the next item, runs the body again, and keeps going until it runs out of items.

## 🧱 The Three Patterns

Nearly every loop you write in this course is one of three patterns (or a mix of them). They all have the same shape:

```text
result = <starting value>          ← set up BEFORE the loop
for item in collection:
    <maybe check something>
    <update result>                ← inside the loop
<use result>                       ← AFTER the loop
```

The only differences are **what you start with** and **how you update it**.

| Pattern        | Starting value | Update                            | You end up with |
|----------------|----------------|-----------------------------------|-----------------|
| **Accumulate** | `0`            | `total += item`                   | a number        |
| **Count**      | `0`            | `count += 1` (when a test passes) | a number        |
| **Filter**     | `[]`           | `.append(item)` (when a test passes) | a **list**   |

Learn these three well. You'll see them again in every notebook for the rest of the course.

### ➕ Pattern 1 — Accumulate
*"Add everything up."*

🔮 Before running, trace `total` on paper for `[87, 91, 74, 88, 95]`:

| pass | `score` | `total` after this pass |
|------|---------|-------------------------|
| start | —      | 0                       |
| 1    | 87      | 87                      |
| 2    | 91      | ?   91                    |
| 3    | 74      | ?   74                    |
| 4    | 88      | ?   88                    |
| 5    | 95      | ?   95                    |

✍️ **My prediction for the final total:**

````python
scores = [87, 91, 74, 88, 95]

total = 0                    # starting value
for score in scores:
    total += score           # update

print("total:", total)
print("average:", total / len(scores))
````

### 🔢 Pattern 2 — Count
*"How many of them pass some test?"*

🔮 How many scores are **80 or higher**?
- 4

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_

````python
count = 0                    # starting value
for score in scores:
    if score >= 80:          # the test
        count += 1           # update: add ONE, not the score

print("scores >= 80:", count)
````

⚠️ Compare this with Accumulate. Accumulate adds `score`. Count adds `1`. Mixing these up is one of the most common beginner bugs.

### 🧹 Pattern 3 — Filter
*"Keep only the ones that pass some test."*

🔮 What list will `high` contain?
- scores = [87, 91, 88, 95]

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_

````python
high = []                    # starting value: an EMPTY list
for score in scores:
    if score >= 80:          # the test
        high.append(score)   # update: keep it

print("high scores:", high)
````

Count and Filter use **the same test**. The only difference is what happens when the test passes: Count adds 1 to a number, and Filter keeps the item in a new list.

## ✏️ Your Turn

A week of daily low temperatures (°F):

```python
temps = [31, 45, 28, 52, 60, 30, 71]
```

Write three **separate** loops:
1. **Accumulate:** the average temperature
2. **Count:** the number of freezing days (32 or below)
3. **Filter:** a list of the days warmer than 50

Before each one, write a comment saying which pattern it is.

````python
temps = [31, 45, 28, 52, 60, 30, 71]
# Your code here

# Accumulate: calculate the average temperature
total = 0
for temp in temps:
    total += temp

average = total / len(temps)
print("Average temperature:", average)


# Count: count the number of freezing days (32 or below)
freezing_days = 0
for temp in temps:
    if temp <= 32:
        freezing_days += 1

print("Freezing days:", freezing_days)


# Filter: create a list of days warmer than 50
warm_days = []
for temp in temps:
    if temp > 50:
        warm_days.append(temp)

print("Days warmer than 50:", warm_days)
````

## 🔥 Challenge (Optional) — A Fourth Pattern: "Best So Far"

How would you find the **largest** score *without* using `max()`?

Idea: start by assuming the first item is the best, then walk through the list. Whenever you see something better, it becomes the new best.

```text
best = <first item>
for item in collection:
    if item is better than best:
        best = item
```

Write it for `scores`. Then answer this: why is starting with `best = 0` a bad idea? (Hint: try it on `[-4, -2, -9]`.)

````python
scores = [87, 91, 74, 88, 95]
# Your code here

best = scores[0]  # start with the first score

for score in scores:
    if score > best:
        best = score

print("Largest score:", best)
````
