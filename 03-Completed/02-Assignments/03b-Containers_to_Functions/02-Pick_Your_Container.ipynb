## 📋 List — "A Bunch of Things"

You already know this one. A list is an **ordered** collection that can **change**. It's the right choice when you have many values of the *same kind* and might add or remove some.

```python
scores = [87, 91, 74, 88]
cart   = ["milk", "eggs", "bread"]
```

## 📍 Tuple — "One Thing With Several Fixed Parts"

**Problem:** You want to store a location. You could use two variables:

````python
lat = 33.91
lon = -98.49
print(lat, lon)
````

But a location isn't two separate things. It's **one** thing with two parts. With two variables, it's easy to pass the latitude somewhere and forget the longitude, or to mix them up.

A tuple keeps the parts together:

````python
location = (33.91, -98.49)   # (latitude, longitude)
color    = (255, 128, 0)     # (red, green, blue)

print(location)
print(color)

# Unpacking: split the parts back out into names
lat, lon = location
print("lat:", lat, "lon:", lon)
````

Why not just use a list? In a tuple, **the position tells you what the part means**. Position 0 of `location` is *always* the latitude. It would make no sense to `.append()` a third number to a GPS coordinate or to `.sort()` an RGB color. Because a tuple can't be changed, nobody can do that by accident.

> A tuple isn't just "an immutable list." It's **a record whose parts have fixed meanings**.

## 🔑 Dictionary — "Look It Up by Name"

**Problem:** You have names and grades in two lists that line up by position (these are called *parallel lists*).

````python
names  = ["Alice", "Bob", "Carol"]
grades = [91, 84, 95]

# What is Bob's grade? First find WHERE Bob is, then use that position.
i = names.index("Bob")
print(grades[i])
````

That works, but it takes two steps, and the two lists can easily get out of sync (sort one and not the other, and every grade now belongs to the wrong person).

What we actually want is to **ask for Bob and get his grade**. That's a dictionary:

````python
grades = {
    "Alice": 91,
    "Bob": 84,
    "Carol": 95
}

print(grades["Bob"])
````

Mental model: each **key** points to its **value**.

```text
"Alice" ───► 91
"Bob"   ───► 84
"Carol" ───► 95
```

You don't care *where* Bob is stored. You ask for him by name.

Looping over a dictionary with `.items()` gives you each key **and** its value:

````python
for name, grade in grades.items():
    print(name, "→", grade)
````

## 🎯 Set — "I Only Care Whether It's There"

**Problem:** A website logged every visit. How many **different** people visited?

````python
visits = ["ana", "ben", "ana", "cam", "ben", "ana"]

unique = set(visits)
print(unique)
print("unique visitors:", len(unique))
print("did dee visit?", "dee" in unique)
````

A set throws away duplicates and doesn't keep things in order. (Your output order may differ from your neighbor's, and that's expected.) Use a set when all you care about is **whether** something is there, not how many times or in what order.

Sets are useful when you compare two groups. Think of a Venn diagram:

````python
terry = {"OOP", "Database", "Algorithms"}
sam   = {"Database", "Networks", "Algorithms"}

print("either   (|):", terry | sam)   # union
print("both     (&):", terry & sam)   # intersection
print("only me  (-):", terry - sam)   # difference
````

```text
  ┌────────────┬──────────────┬────────────┐
  │    OOP     │   Database   │  Networks  │
  │            │  Algorithms  │            │
  └────────────┴──────────────┴────────────┘
   terry - sam   terry & sam    sam - terry
  └───────── terry | sam (everything) ─────┘
```

## 🪆 Nested Containers — Taking Expressions Apart

Real data is usually containers **inside** containers. The most common shape in this course (and in Pandas later on) is a **list of dictionaries**, where each dictionary is one row of a table:

````python
students = [
    {"name": "Alice", "grade": 91},
    {"name": "Bob",   "grade": 84},
    {"name": "Carol", "grade": 95}
]
````

🔮 **Don't run anything yet.** What is `students[1]["grade"]`?

Don't try to answer in one jump. Evaluate it **one step at a time, left to right**. Each step gives you a value, and the next step works on that value:

```text
students[1]["grade"]
└────┬────┘
     ▼
{"name": "Bob", "grade": 84}["grade"]      ← step 1: students[1] is a dictionary
                └───┬───┘
                    ▼
                    84                     ← step 2: look up "grade" in THAT dictionary
```

Now try these on paper using the same step-by-step method. Write the value after **each** step:

```python
students[0]    # 'name': 'Alice', 'grade': 91
students[0]["name"] #Alice
students[-1]["name"] # Carol
students[-1]["name"][0] # C
len(students) # 3
len(students[2]) #2
```

✍️ **My prediction:** _(write it here BEFORE you run the next cell)_

````python
print(students[0])
print(students[0]["name"])
print(students[-1]["name"])
print(students[-1]["name"][0])
print(len(students))
print(len(students[2]))
````

The same method works when the nesting goes the other way (a dictionary holding a list):

````python
course = {
    "title": "CMPS 3603",
    "scores": [88, 72, 95]
}

# 🔮 Predict each before running!
print(course["scores"]) #[88, 72, 95]
print(course["scores"][1]) #72
print(len(course["scores"])) #3
print(course["title"][-4:]) # none
````

### 🧱 The Same Three Patterns on Nested Data

A list of dictionaries is still a list, so the loop works the same way. The only change is that each item is now a dictionary, so you reach inside it with `["grade"]`.

````python
students = [
    {"name": "Alice", "grade": 91},
    {"name": "Bob",   "grade": 84},
    {"name": "Carol", "grade": 95}
]

# Accumulate
total = 0
for s in students:
    total += s["grade"]
print("average:", total / len(students))

# Count
count = 0
for s in students:
    if s["grade"] >= 90:
        count += 1
print("A students:", count)

# Filter (keep just the names)
honor_roll = []
for s in students:
    if s["grade"] >= 90:
        honor_roll.append(s["name"])
print("honor roll:", honor_roll)
````

## 🔥 Challenge (Optional) — Counting With a Dictionary

Remember "Word → number of times it appears" from Pick Your Weapon? Build it. Start with an empty dictionary and walk the list. If the word is already a key, add 1. If it isn't, set it to 1.

```python
words = ["red", "blue", "red", "green", "blue", "red"]
# goal: {'red': 3, 'blue': 2, 'green': 1}
```

This is a **Count** pattern that keeps a separate counter for every word. You'll see it again in the functions notebooks.

````python
words = ["red", "blue", "red", "green", "blue", "red"]
counts = {}
# Your loop here
for word in words:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1


print(counts)
````
