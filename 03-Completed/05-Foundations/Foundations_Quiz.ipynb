## Q5. Files and Paths (05)
Use `pathlib` and `with open(..., encoding='utf-8')`:
1. Write `Hello World` to `OUT / 'hello.txt'`.
2. **Append** a second line, `Goodbye World`, without erasing the first.
3. Read the whole file back and print it.
4. Print the file's `.name`, `.suffix`, and whether it `.exists()`.

```python
# Your code here
# File path
file_path = OUT / "hello.txt"

# 1. Write "Hello World"
with open(file_path, "w", encoding="utf-8") as f:
    f.write("Hello World\n")

# 2. Append a second line
with open(file_path, "a", encoding="utf-8") as f:
    f.write("Goodbye World\n")

# 3. Read the whole file and print it
with open(file_path, "r", encoding="utf-8") as f:
    content = f.read()

print(content)

# 4. Print file information
print("Name:", file_path.name)
print("Suffix:", file_path.suffix)
print("Exists:", file_path.exists())

```

## Q6. JSON (05)
Create a dictionary describing a course (name, number of credits, and a **list** of topics). Save it to `OUT / 'course.json'` with `indent=2`, read it back into a new variable, and print whether the two dictionaries are equal (`==`).

```python
# Your code here
# Create a course dictionary
course = {
    "name": "Introduction to Python",
    "credits": 3,
    "topics": ["Variables", "Loops", "Functions", "File Handling"]
}

# Save to JSON with indent=2
json_file = OUT / "course.json"

with open(json_file, "w", encoding="utf-8") as f:
    json.dump(course, f, indent=2)

# Read it back into a new variable
with open(json_file, "r", encoding="utf-8") as f:
    course_loaded = json.load(f)

# Check whether the dictionaries are equal
print(course == course_loaded)
```

## Q9. Choosing and Building a Chart (07)
Generate 200 exam scores with `np.random.seed(1)` and `np.random.normal(70, 12, size=200)`. Pick the **best chart** for showing their distribution, and draw it with a title, axis labels, and a dashed vertical line marking the mean (with a legend).

```python
# Your code here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
scores = np.random.normal(70, 12, size=200)

# Create DataFrame
df_scores = pd.DataFrame({"Scores": scores})

# Calculate mean
mean_score = df_scores["Scores"].mean()

# Plot histogram
plt.figure(figsize=(10, 4))
plt.hist(
    df_scores["Scores"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)

# Add mean line
plt.axvline(
    mean_score,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_score:.2f}"
)

# Labels and title
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.7)
```

In one or two sentences: why did you choose that chart type over a bar chart or a line plot?

*Your Answer:*For showing the distribution of exam scores, the best chart is a histogram because it groups the scores into ranges (bins) and shows how frequently scores occur in each range.

## Q10. Timing (08)
Compare squaring one million numbers with a Python loop (or list comprehension) over a **list** and with NumPy on an **array**:
1. Create both inputs **before** timing, so the setup isn't measured.
2. Confirm that both approaches produce the same values.
3. Use `%timeit -o` for each, and print how many times faster NumPy is.

```python
# Your code here
import numpy as np

# Create inputs BEFORE timing
numbers = list(range(1_000_000))
arr = np.arange(1_000_000)

# Confirm both produce the same values
python_result = [x**2 for x in numbers]
numpy_result = arr**2

print("Results equal:", np.array_equal(np.array(python_result), numpy_result))

# Time each approach
python_time = %timeit -o [x**2 for x in numbers]
numpy_time = %timeit -o arr**2

# Calculate speedup
speedup = python_time.average / numpy_time.average

print(f"\nPython average time: {python_time.average:.6f} seconds")
print(f"NumPy average time:  {numpy_time.average:.6f} seconds")
print(f"NumPy is approximately {speedup:.1f}x faster.")
```
