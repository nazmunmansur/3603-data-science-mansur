# 🐼 01 - Pandas Series and DataFrames

Time to meet the two most essential data structures in data science:
- `Series`: a one-dimensional labeled array (like a list, but every value has a label)
- `DataFrame`: a two-dimensional table with labeled rows and columns

## ✅ Learning Goals
- Create a `Series` from a list and from a dictionary
- Inspect a `Series` and do math and filtering on all of its values at once
- Create a `DataFrame` from a dictionary of columns, and from a **list of dictionaries** (one per row)
- Access columns and rows of a DataFrame
- Inspect a DataFrame's shape, columns, and summary info

````python
import pandas as pd
````

## 🧩 The Two Central Pandas Objects

| Object | Meaning | Example |
|---|---|---|
| `Series` | **One** labeled column of values | Product prices |
| `DataFrame` | A labeled **table** (rows and columns) | A product catalog |

```text
Series:            DataFrame:
0    19.99            product   price  stock
1    24.99         0  Notebook   4.99     32
2    12.50         1  Pen        1.50    120
```

A DataFrame is really a set of Series that share the same row labels (the **index**). Each column is one Series.

## 🔢 Series from a List
Simple but powerful — the `index` argument gives each value a label.

````python
prices = pd.Series([2.99, 4.49, 1.99], index=['Apple', 'Milk', 'Bread'])
print('Grocery Prices:')
print(prices)

print('\nPrice of Milk:', prices['Milk'])
````

Leave out `index=`, and pandas numbers the values `0, 1, 2, ...`, the **default index**. The labels are on the left, and the values are on the right:

````python
scores = pd.Series([82, 91, 76, 88, 95])
scores
````

## 🗺️ Series from a Dictionary
The dictionary's keys automatically become the Series' index.

````python
population = {
    'Texas': 29_000_000,
    'California': 39_000_000,
    'New York': 19_000_000,
}

state_pop = pd.Series(population)
print('State Populations:')
print(state_pop)

# Series support math, applied to every value at once
print('\nPopulation in millions:')
print(state_pop / 1_000_000)
````

## 🧮 Working with a Series
A Series knows its size and type, and it can summarize itself:

````python
print('shape:', scores.shape)   # (5,): 5 values, one dimension
print('dtype:', scores.dtype)
print('mean: ', scores.mean())
print('min/max:', scores.min(), scores.max())
````

````python
scores.describe()
````

Like a NumPy array (Foundations notebook 07), math applies to **every** value at once, and a condition gives you a True/False **mask** that you can use to filter:

````python
print(scores + 5)              # curve every score
print(scores[scores >= 85])    # keep only the scores that pass
````

To pick out several labels at once, pass a **list** of labels. Note the double brackets:

````python
prices[['Apple', 'Bread']]
````

## 📋 Creating a DataFrame
A dictionary of columns becomes a table — each key is a column name, each value is a list of that column's data.

````python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'GPA': [3.9, 2.7, 3.4, 3.8],
    'Credits': [90, 45, 60, 120],
    'Graduating': [False, False, False, True],
}

students = pd.DataFrame(data)
students
````

Every list in the dictionary must be the **same length**. Each one is a column, and a table can't have a column that's shorter than the others.

### 📋 From a List of Dictionaries
In Modules 03 and 03b, you stored table rows as a **list of dictionaries**, one dictionary per row. That's also the shape most JSON files and web APIs give you. Hand that list to `pd.DataFrame`, and each dictionary becomes a row:

````python
purchases = pd.DataFrame([
    {'purchase_id': 1, 'buyer': 'Avery',  'amount': 25.00},
    {'purchase_id': 2, 'buyer': 'Jordan', 'amount': 42.50},
    {'purchase_id': 3, 'buyer': 'Avery',  'amount': 18.75},
])
purchases
````

| You have... | Use | Each item becomes... |
|---|---|---|
| A dictionary of **columns** (`{'GPA': [...], ...}`) | `pd.DataFrame(dict_of_lists)` | a column |
| A list of **rows** (`[{...}, {...}]`) | `pd.DataFrame(list_of_dicts)` | a row |

## 🎯 Column Access and Row Access

````python
# Access a column
print(students['GPA'])
````

````python
# Access a row by label with .loc[]
print(students.loc[2])
````

Notebook 03 covers selecting rows and columns in depth, including the difference between `.loc[]` and `.iloc[]`.

## 🔥 Challenge (Optional) — Series Alignment
When you do math between two `Series`, Pandas lines up values by their **index labels**, not their position — even if they're in a different order or missing entries. Try this:
```python
a = pd.Series({'x': 1, 'y': 2, 'z': 3})
b = pd.Series({'z': 10, 'x': 20})
print(a + b)
```
Notice `'y'` becomes `NaN` — there was nothing in `b` to add to it. This label-based alignment (not position-based) is one of the biggest differences between a Pandas `Series` and a plain Python list.

````python
a = pd.Series({'x': 1, 'y': 2, 'z': 3})
b = pd.Series({'z': 10, 'x': 20})
print(a + b)
````
