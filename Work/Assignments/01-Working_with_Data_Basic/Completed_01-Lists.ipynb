# 📘 01 - Python Lists

Explore how to use lists to store and manipulate data.

## ✅ Learning Goals
- Create and modify Python lists
- Access elements using indexing
- Use list methods like `append()`, `remove()`, and `sort()`

````python
# 🔧 Create a List
my_list = [10, 20, 30, 40, 50]
print(my_list)
````

````python
# 🎯 Access Elements
print('First item:', my_list[0])
print('Last item:', my_list[-1])
````

````python
# ➕ Modify List
my_list.append(60)
my_list.remove(30)
print(my_list)
````

````python
# 🔁 Loop Through List
for item in my_list:
    print(item)
````

### Practice 2 — List Methods (medium)
Starting from `numbers = [5, 3, 8, 1, 9, 2]`:
- Sort the list in place
- Reverse it
- Print the length

Expected final output: `[9, 8, 5, 3, 2, 1]` and `6`.

````python
numbers = [5, 3, 8, 1, 9, 2]
# Sort the list
Sort = numbers.sort()
print(numbers)
# Reserve the list
Reserve = numbers.reverse()
print(numbers)
#Length of the list
print(len(numbers))
````

### Practice 3 — Filtering (harder)
Given `scores = [55, 72, 90, 43, 88, 67]`, use a loop and an `if` statement to build a new list called `passing` containing only the scores 60 and above. Print `passing`.

````python
scores = [55, 72, 90, 43, 88, 67]
passing = []
# Your loop here
for score in scores:
    if score >= 60:
        passing.append(score)
print(passing)
````

## 🔥 Challenge (Optional)
Want to build a list in one line? Python has *list comprehensions* — a compact way to write a loop that builds a list.

```python
squares = [n**2 for n in range(10)]
```

Try it: build a list of the **even** numbers from 1 to 20 using a list comprehension (hint: add an `if` at the end).

````python
# Your list comprehension here
evens = [n for n in range(1, 21) if n % 2 == 0]
print(evens)
````
