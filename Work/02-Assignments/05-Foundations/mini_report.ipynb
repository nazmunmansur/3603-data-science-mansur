## Study Time Report

# Question:
- "Which day this week did I spend the most time studying?"

```python
# Imports and setup
import matplotlib.pyplot as plt
```

# The study data
The data using here is the last weeek data for reviewing the study hours depending on the days. 'days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]' the days variable is taking all the weekdays and the hours is representing the number of hours.

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
hours = [2, 3, 1, 4, 2]
```

# Analysis:
Our goal is to fing out the maximum hours of study and which day you get to do the maximum study hours.

```python
max_hours = max(hours)
max_day = days[hours.index(max_hours)]

print(f"The day with the most study time was {max_day} with {max_hours} hours.")
```

# Visulization

```python
import matplotlib.pyplot as plt
plt.plot(days, hours, marker="o")
plt.title("Study Hours by Day")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.show()
```

# Finding and Limitation
Based on the data, Thursday had the highest amount of study time with 4 hours.

This example uses made-up data and only covers one week, so the results do not represent long-term study habits.
