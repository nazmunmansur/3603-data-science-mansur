## 2. Line Plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)')
plt.title('Line Plot Example')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.show()
```

## 3. Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = 300 * np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, alpha=0.6, cmap='viridis')
plt.title('Scatter Plot Example')
plt.xlabel('x')
plt.ylabel('y')
plt.colorbar(label='color scale')
plt.show()
```

## 4. Bar Plot

## 5. Customizing Plots

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)', linestyle='--', color='blue')
plt.plot(x, y2, label='cos(x)', linestyle=':', color='red')
plt.title('Customized Line Plot')
plt.xlabel('x (radians)')
plt.ylabel('value')
plt.legend()
plt.grid(True)
plt.show()
```

✅ **Your Turn**: Customize your own plot with labels, a legend, and grid lines.

```python
# Your turn here
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1,
         label='sin(x)',
         linestyle='-',
         color='pink',
         linewidth=2)

plt.plot(x, y2,
         label='cos(x)',
         linestyle='-.',
         color='red',
         linewidth=2)

plt.title('Customized Line Plot')
plt.xlabel('Angle (radians)')
plt.ylabel('Function Value')

plt.legend(loc='upper right')

# Customize grid
plt.grid(
    True,
    linestyle='-.',
    linewidth=1,
    color='gray',
    alpha=0.7
)

plt.show()
```

### Practice 2 — Two Lines, One Legend (medium)
Plot `y = x` and `y = x**2` on the same axes for `x = np.linspace(0, 5, 50)`, with a legend distinguishing them and a title.

```python
# Your code here
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 5, 50)
y1 = x
y2 = x**2

plt.plot(x, y1,
         label='y1',
         linestyle='-',
         color='pink',
         linewidth=2)

plt.plot(x, y2,
         label='y2',
         linestyle='-.',
         color='red',
         linewidth=2)

plt.title('Customized Line Plot')
plt.xlabel('x values')
plt.ylabel('y1 = x and y2 = x**2')

plt.legend(loc='upper right')

# Customize grid
plt.grid(
    True,
    linestyle='-.',
    linewidth=1,
    color='gray',
    alpha=0.7
)

plt.show()
```

### Practice 3 — Multiple Subplots (harder)
Use `plt.subplots(1, 2)` to create two plots side by side: a line plot of `sin(x)` on the left, and a bar plot of `categories`/`values` (from section 4) on the right.

```python
# Your code here
import numpy as np
import matplotlib.pyplot as plt

# Data for line plot
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Data for bar plot
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

# Create 1 row, 2 columns of subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Left subplot: sin(x)
axes[0].plot(x, y, color='blue', label='sin(x)')
axes[0].set_title('Sine Function')
axes[0].set_xlabel('x (radians)')
axes[0].set_ylabel('sin(x)')
axes[0].legend()
axes[0].grid(True)

# Right subplot: bar chart
axes[1].bar(categories, values, color='orange')
axes[1].set_title('Bar Plot Example')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Value')

# Adjust spacing
plt.tight_layout()

# Display both plots
plt.show()
```
