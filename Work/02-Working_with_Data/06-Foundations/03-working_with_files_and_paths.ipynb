## 3. JSON Files

````python
import json

data = {'name': 'Graylian', 'type': 'mascot', 'skills': ['howl', 'parrot talk']}

# Write JSON
with open('data.json', 'w') as f:
    json.dump(data, f)

# Read JSON
with open('data.json', 'r') as f:
    loaded = json.load(f)
loaded
````

## 4. CSV Files

````python
import csv

# Writing CSV
rows = [['Name', 'Score'], ['Alice', 90], ['Bob', 85]]
with open('scores.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open('scores.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

Open `example.txt` in append mode (`'a'` instead of `'w'`) and add one more line to it without erasing what's already there. Read the whole file back to confirm.

### Practice 2 — JSON Round Trip with a List (medium)
Save a list of 3 dictionaries (each with `'name'` and `'score'` keys) to `results.json`, then read it back and print the average score.

````python
# Your code here
import json

# List of dictionaries
results = [
    {"name": "Ana", "score": 92},
    {"name": "Ben", "score": 78},
    {"name": "Cy", "score": 85}
]

# Save to results.json
with open("results.json", "w") as file:
    json.dump(results, file, indent=4)

# Read from results.json
with open("results.json", "r") as file:
    data = json.load(file)

# Calculate average score
average = sum(student["score"] for student in data) / len(data)

print("Average score:", average)
````

### Practice 3 — CSV to Dictionaries (harder)
Read `scores.csv` back in using `csv.DictReader` instead of `csv.reader`, and print each row as `'Name got a Score'` using the column names as dictionary keys.

````python
# Your code here
with open('scores.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['Name']} got a {row['Score']}")
````

## 🔥 Challenge (Optional) — Paths That Work Anywhere
Typing `'example.txt'` only works because it happens to be in the current directory. The `pathlib` module (built into Python) gives you a more robust way to build paths that works the same on Windows, Mac, and Linux.
```python
from pathlib import Path

data_folder = Path('data')
file_path = data_folder / 'scores.csv'
print(file_path)
print(file_path.exists())
```
