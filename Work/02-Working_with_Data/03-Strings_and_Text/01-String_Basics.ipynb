# 📘 01 - String Basics

Strings are how Python represents text — and they behave a lot like the lists and tuples you already know.

## ✅ Learning Goals
- Create strings with single, double, and triple quotes
- Understand why Python doesn't distinguish `'...'` from `"..."`
- Use escape characters and multi-line strings
- Index and slice strings like you did with lists and tuples
- Understand why strings are immutable

## 🔤 Creating Strings

````python
greeting = 'Hello'
name = "Ada"
print(greeting, name)
print(type(greeting))
````

### Single vs. Double Quotes — No Difference
Python doesn't care which one you use — `'...'` and `"..."` create the exact same type of object. The only reason to pick one over the other is to avoid escaping a quote character that appears *inside* the string.

````python
quote1 = "It's a nice day"      # double quotes let an apostrophe sit unescaped
quote2 = 'She said "hello"'      # single quotes let double quotes sit unescaped
print(quote1)
print(quote2)
````

### Escape Characters
If you need both kinds of quotes in the same string (or a literal backslash), escape the character with a backslash `\`.

````python
escaped = 'She said \'hi\' to me'
tabbed = 'Name:\tAda'
newlined = 'Line one\nLine two'
print(escaped)
print(tabbed)
print(newlined)
````

### Triple-Quoted Strings
Wrap text in `'''` or `"""` to write a string across multiple lines without needing `\n` everywhere.

````python
bio = """Ada Lovelace
Mathematician
Wrote the first algorithm intended for a machine."""
print(bio)
````

## 🔍 Indexing and Slicing
Strings are sequences too — the same indexing and slicing rules from Lists and Tuples apply here.

````python
word = 'Python'
print(word[0])
print(word[-1])
print(word[0:3])
print(word[::-1])  # step of -1 reverses the string
````

## 🚫 Strings Are Immutable
Just like tuples, once a string is created you can't change one character in place — you have to build a new string instead.

````python
word = 'Python'
# word[0] = 'J'  # Uncomment to see the error
new_word = 'J' + word[1:]
print(new_word)
````

## ➕ Concatenation and Repetition

````python
a = 'Py'
b = 'thon'
print(a + b)
print(a * 3)
````
