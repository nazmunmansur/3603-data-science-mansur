# 📘 03 - Python Dictionaries

Dictionaries store key-value pairs for fast lookups.

## ✅ Learning Goals
- Create and access dictionary items
- Use dictionary methods like `.get()` and `.items()`

````python
# 📚 Create a Dictionary
student = {'name': 'Alex', 'age': 21, 'major': 'CS'}
print(student)
````

````python
# 🔍 Access Values
print(student['name'])
print(student.get('gpa', 'N/A'))
````

````python
# 🔁 Loop Over Dictionary
for key, value in student.items():
    print(key, '->', value)
````

### Practice 1 — Update, Add, Remove (easy)
Starting from `student = {'name': 'Alex', 'age': 21, 'major': 'CS'}`:

### Practice 2 — Build a Dictionary from Two Lists (medium)
Given `names = ['Ana', 'Ben', 'Cy']` and `scores = [90, 85, 77]`, build a dictionary that maps each name to their score. (Hint: loop with `zip()`, or look up `dict(zip(...))` — remember tuples from the last notebook?)

````python
names = ['Ana', 'Ben', 'Cy']
scores = [90, 85, 77]
# Your dictionary-building here
student = dict(zip(names, scores))
print(student)
````

### Practice 3 — Nested Dictionaries (harder)
```python
students = {
    'Ana': {'age': 20, 'major': 'CS'},
    'Ben': {'age': 22, 'major': 'History'},
}
```
Loop through `students` and print each name with their age, like `Ana is 20 years old`.

````python
students = {
    'Ana': {'age': 20, 'major': 'CS'},
    'Ben': {'age': 22, 'major': 'History'},
}
# Your loop here
for name, info in students.items():
    print(name, '->', info) 
````

## 🔥 Challenge (Optional)
Dictionaries are great for counting things. Try building a **word frequency counter**:

```python
text = 'the quick brown fox jumps over the lazy dog the fox runs'
words = text.split()
```

Count how many times each word appears using a dictionary (hint: `counts[word] = counts.get(word, 0) + 1`).

For a step further, look up `collections.Counter` — it does the same thing in one line.

````python
# Your word counter here
text = 'the quick brown fox jumps over the lazy dog the fox runs'
words = text.split()
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1
print(counts)
````
