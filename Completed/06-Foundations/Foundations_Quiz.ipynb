## Q4. File I/O
Write Python code that saves `Hello World` to a file called `hello.txt`, then reads it back and prints the contents.

```python
# Write "Hello World" to hello.txt
with open("hello.txt", "w") as file:
    file.write("Hello World")

# Read the contents back from hello.txt
with open("hello.txt", "r") as file:
    contents = file.read()

# Print the contents
print(contents)
```

## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 11)
y = x**2

plt.plot(x, y,
         label='y',
         linestyle='-.',
         color='red',
         linewidth=2)

plt.title('Customized Line Plot')
plt.xlabel('x values')
plt.ylabel('y = x**2')

plt.legend(loc='upper right')

# Customize grid
plt.grid(
    True,
    linestyle='-.',
    linewidth=1,
    color='gray',
    alpha=0.7
)

plt.show()
```
