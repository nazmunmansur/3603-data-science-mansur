## 4. Writing and Running Files
`%%writefile` (a **cell** magic — note the double `%%`) saves the entire cell's contents to a file instead of running it as Python.

```python
%%writefile hello.py
print('Hello from a file!')
```

```python
%run hello.py
```

## 5. Output and Formatting

```python
%%capture cap
print('This will not display right away')
```

```python
cap.show()
```

```python
%%html
<h1 style="color: maroon">Magics are Magical ✨</h1>
```

## 6. Discover More
`%lsmagic` lists every magic command available. Adding `?` after any of them (including a regular function) shows its docstring.

```python
%lsmagic
```

```python
%timeit?
```
