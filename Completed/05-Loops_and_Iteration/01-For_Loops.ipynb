# 📘 01 - For Loops

Learn how to use `for` loops to iterate over sequences.

## ✅ Learning Goals
- Use `for` with lists, strings, and `range()`
- Get the index and value together with `enumerate()`
- Loop over two sequences together with `zip()`
- Write nested loops
- If you know C++: see how Python's for-each differs from index-based loops

## 🔁 Loop Over a List

````python
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)
````

## 🔤 Loop Over Characters in a String

````python
word = 'hello'
for char in word:
    print(char)
````

## 🔢 Using `range()`

````python
for i in range(5):
    print(i)
````

## ➕ Loop With Math

````python
for i in range(1, 6):
    print(f'{i} squared is', i ** 2)
````

### 🔀 Cross-Language Note: C++ For Loops
C++'s classic `for` loop makes you manage an index yourself. Python's `for x in sequence:` is a **for-each** loop — it hands you the item directly, no index bookkeeping required.
```cpp
// C++
vector<string> fruits = {"apple", "banana", "cherry"};
for (int i = 0; i < fruits.size(); i++) {
    cout << fruits[i] << endl;
}
```
```python
# Python — no index management needed
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)
```
When you *do* need the index in Python, don't manage a counter by hand — reach for `enumerate()` instead, below.

## 🔢 `enumerate()` — When You Need the Index Too

````python
fruits = ['apple', 'banana', 'cherry']
for i, fruit in enumerate(fruits):
    print(i, fruit)
````

## 🔗 `zip()` Recap — Looping Over Two Lists Together
You saw `zip()` back in the Tuples module. It's especially handy inside a `for` loop, when you have two related lists you want to walk through together.

````python
names = ['Ana', 'Ben', 'Cy']
scores = [90, 85, 77]
for name, score in zip(names, scores):
    print(f'{name}: {score}')
````

## 🔁 Nested Loops
A loop can contain another loop. The inner loop runs completely for *each* pass of the outer loop.

````python
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end=' ')
    print()
````

### Practice 2 — Numbered List with enumerate() (medium)
Given `groceries = ['milk', 'eggs', 'bread']`, use `enumerate()` to print `'Item 1: milk'`, `'Item 2: eggs'`, etc. (hint: `enumerate()` starts at 0 by default — you'll need to adjust).

````python
groceries = ['milk', 'eggs', 'bread']
# Your enumerate() loop here
for i, item in enumerate(groceries):
    print(f'Item {i + 1}: {item}')
````

### Practice 3 — Report from Two Lists (harder)
Given `students = ['Ana', 'Ben', 'Cy']` and `grades = [92, 78, 85]`, use `zip()` to print a line per student like `Ana: 92 (A)` — include the letter grade (90+ A, 80-89 B, else C) by combining this with an `if`/`elif`/`else`.

````python
students = ['Ana', 'Ben', 'Cy']
grades = [92, 78, 85]

# Your zip() loop here
for student, grade,  in zip(students, grades):
    if grade >= 90:
        letter = 'A'
    elif grade >= 80:
        letter = 'B'
    else:
        letter = 'C'
    print(f'{student}: {grade} ({letter})')
````
