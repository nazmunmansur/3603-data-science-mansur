# 📘 03 - File Loops and `with`

Use `with open(...)` to read files and process their contents, line by line.

## ✅ Learning Goals
- Open and read files safely using `with`
- Loop over lines in a file
- Use `.strip()` and `.split()` to process text
- Handle a missing file gracefully with `try`/`except`
- If you know C++: see how `with` compares to RAII

## ✍️ First, Create a Sample File
File reading needs something to read! Run this cell once — it creates `sample.txt` in the same folder as this notebook, which the rest of the notebook uses.

````python
lines = [
    'Ana,22,CS',
    'This line has more than three words in it',
    'Ben,19,History',
    'Short line',
    'error: could not load dataset',
    'Cy,25,Art',
    'This is another longer line with several words',
]

with open('sample.txt', 'w') as f:
    for line in lines:
        f.write(line + '\n')

print('sample.txt created!')
````

## 📝 Reading a File

````python
with open('sample.txt', 'r') as file:
    for line in file:
        print(line.strip())
````

## 🔍 Processing Each Line
Real-world text data is messy — not every line will be in the format you expect. Notice that lines without a comma just come back as a one-item list; that's not an error, just something to be aware of.

````python
with open('sample.txt', 'r') as f:
    for line in f:
        parts = line.strip().split(',')
        print('Parts:', parts)
````

## 🚦 Handling a Missing File Safely
Remember `try`/`except` from the Functions module (the duck-typing example)? File reading is one of the most common places you'll actually use it — the file you're trying to open might not exist.

````python
try:
    with open('does_not_exist.txt', 'r') as f:
        print(f.read())
except FileNotFoundError:
    print('That file does not exist yet — check the filename and path.')
````

### 🔀 Cross-Language Note: `with` vs. C++ RAII
If you've used C++, `with` plays a similar role to RAII (Resource Acquisition Is Initialization) with something like `ifstream` — the file gets closed automatically when the block ends (or even if an error happens inside it), instead of you having to remember to call `.close()` yourself.
```cpp
// C++ — ifstream closes automatically when it goes out of scope (RAII)
ifstream file("sample.txt");
string line;
while (getline(file, line)) {
    cout << line << endl;
}
// file closes automatically here
```
```python
# Python — "with" guarantees the file closes, even if an error happens inside
with open('sample.txt', 'r') as file:
    for line in file:
        print(line.strip())
# file is already closed here
```

### Practice 2 — Long Lines Only (medium)
Loop over `sample.txt` and print only the lines with **more than 3 words** (hint: `len(line.split()) > 3`).

````python
# Your loop here
with open('sample.txt', 'r') as file:
    for line in file:
        if len(line.split()) > 3:
            print(line.strip())

````

### Practice 3 — Line and Word Counts (harder)
Loop over `sample.txt` once and count: the total number of lines, and the total number of words across the whole file. Print both counts.

````python
# Your loop here
line_count = 0
word_count = 0

with open("sample.txt", "r") as file:
    for line in file:
        line_count += 1
        word_count += len(line.split())

print("Total lines:", line_count)
print("Total words:", word_count)
````

````python
# Read sample.txt and collect error lines
error_lines = []

with open("sample.txt", "r") as file:
    for line in file:
        if "error" in line:
            error_lines.append(line)

# Write error lines to filtered.txt
with open("filtered.txt", "w") as file:
    file.writelines(error_lines)

# Open filtered.txt and print its contents
with open("filtered.txt", "r") as file:
    for line in file:
        print(line.strip())
````
