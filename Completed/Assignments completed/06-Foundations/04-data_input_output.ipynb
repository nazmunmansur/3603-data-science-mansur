```python
import pandas as pd

def load_data(filename):
    try:
        df = pd.read_csv(filename)
        print(f"Successfully loaded {filename}")
        return df
    except FileNotFoundError:
        print(f"Sorry, '{filename}' was not found.")
        return None

# Test with an existing file
load_data('students.csv')

# Test with a file that does not exist
load_data('does_not_exist.csv')

try:
    with open('does_not_exist.txt', 'r') as f:
        print(f.read())
except FileNotFoundError:
    print('That file does not exist yet — check the filename and path.')
```
