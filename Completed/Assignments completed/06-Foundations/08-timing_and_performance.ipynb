# ⏱️ 08 - Timing and Performance

In data science, performance matters. Some code runs fast, some slow. Jupyter/IPython gives you tools to measure runtime easily.

## ✅ Learning Goals
- Use `%time` and `%timeit` to measure single expressions
- Use `%%time` and `%%timeit` to measure a whole cell
- Compare loops, list comprehensions, and NumPy for the same task
- Understand why measuring beats guessing when optimizing code

## 1. `%time`
`%time` runs an expression **once** and reports how long it took.

```python
%time sum(range(1_000_000))
```

## 2. `%timeit`
`%timeit` runs an expression **many times** and reports an average — much more reliable than a single `%time` measurement, since one run can be thrown off by what else your computer is doing at that moment.

```python
numbers = list(range(1_000))
%timeit [x**2 for x in numbers]
```

## 3. `%%time` and `%%timeit` for a Whole Cell
Put `%%time` or `%%timeit` as the **first line** of a cell to measure everything in it, not just one expression.

```python
%%timeit
total = 0
for i in range(1_000_000):
    total += i
```

## 4. Comparing Loops vs. NumPy
This is the payoff: the same calculation, three different ways, timed side by side. You'll learn NumPy properly soon — for now, just notice the difference `numbers**2` (a **vectorized** operation) makes.

```python
import numpy as np

numbers = np.arange(1_000_000)

# Python loop (as a list comprehension)
%timeit [x**2 for x in numbers]

# NumPy vectorized
%timeit numbers**2
```

## 5. Why This Matters
- Performance differences become huge with large datasets.
- Vectorized operations (NumPy, Pandas) are usually far faster than a Python-level loop.
- `%timeit` is how you find out *before* you commit to an approach — don't guess, measure.

## 🔥 Challenge (Optional) — Where's the Line?
For very small inputs, a plain Python loop can actually be just as fast as (or faster than) NumPy, because NumPy has a small fixed overhead per call. Find roughly how small an array has to be before the loop wins, by timing both approaches at a few different sizes (e.g. 10, 100, 1,000, 100,000).

```python
import numpy as np

sizes = [10, 100, 1_000, 10_000, 100_000]

for size in sizes:
    numbers = list(range(size))
    arr = np.arange(size)

    print(f"\nSize = {size}")

    # Python loop
    %timeit -o [x**2 for x in numbers]

    # NumPy
    %timeit -o arr**2
```

```python
# Your investigation here
# For very small arrays (around 10 to a few dozen elements),
# a Python list comprehension can be as fast as or faster than NumPy
# because NumPy has a fixed overhead for creating and managing arrays.
#
# As the array size increases (hundreds or thousands of elements),
# NumPy becomes significantly faster and the performance gap grows quickly.
```
