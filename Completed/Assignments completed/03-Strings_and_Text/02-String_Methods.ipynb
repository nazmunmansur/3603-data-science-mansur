# 📘 02 - String Methods

Python strings come with a huge set of built-in methods for cleaning, searching, and transforming text.

## ✅ Learning Goals
- Change case with `.upper()`, `.lower()`, `.title()`
- Clean up whitespace with `.strip()` and friends
- Break apart and rebuild strings with `.split()` and `.join()`
- Search and replace with `.find()`, `in`, and `.replace()`
- Ask yes/no questions about a string with boolean check methods

## 🔠 Case Methods

```python
name = 'ada LOVELACE'
print(name.upper())
print(name.lower())
print(name.title())
```

## ✂️ Whitespace Methods
`repr()` below just makes the spaces visible in the output — normally you'd use `print()` alone.

```python
messy = '   lots of space   '
print(repr(messy.strip()))
print(repr(messy.lstrip()))
print(repr(messy.rstrip()))
```

## 🔗 split() and join()
`.split()` breaks a string into a list of pieces. `.join()` does the reverse — it glues a list of strings back together with a separator.

```python
sentence = 'the quick brown fox'
words = sentence.split()
print(words)
print('-'.join(words))
```

## 🔎 Searching and Replacing

```python
text = 'data science is fun'
print(text.find('science'))   # index where it starts, or -1 if not found
print('science' in text)       # simpler yes/no membership check
print(text.replace('fun', 'powerful'))
```

## ❓ Boolean Check Methods
Remember the password-strength example from Control Flow, where you checked `char.isdigit()`? That's one of many string methods that ask a yes/no question and return a `bool`.

```python
print('12345'.isdigit())
print('hello'.isalpha())
print('hello.py'.endswith('.py'))
print('Hello World'.startswith('Hello'))
```

## 🔥 Challenge (Optional) — Build a Simple Validator
Combine several boolean check methods to validate a username: it should contain **only lowercase letters and digits**, and must **not start with a digit**.

Test your validator against `'ada99'` (valid) and `'99ada'` (invalid).

```python
def is_valid_username(username):
    # Your validation logic here
    return (
            username.isalnum() and      # only letters and digits
            username.islower() and      # all letters are lowercase
            not username[0].isdigit()   # does not start with a digit
        )
print(is_valid_username('ada99'))
print(is_valid_username('99ada'))
```
